```python
import pandas as pd
import numpy as np

np.random.seed(42)
n=500

data = ({
    'CustomerID':range(1, 501),
    'Age': np.random.randint(22, 65, n),
    'Gender':np.random.choice(['Male', 'Female'], n),
    'Tenure':np.random.randint(2, 60, n),
    'Usage Frequency':np.random.randint(3, 30, n),
    'Support Calls':np.random.randint(0, 10, n),
    'Payment Delay':np.random.randint(2, 30, n),
    'Subscription Type':np.random.choice(['Basic', 'Standard', 'Premium'], n),
    'Contract Length':np.random.choice(['Monthly', 'Annual', 'Quarterly'], n),
    'Total Spend':np.random.randint(100, 1000, n),
    'Last Interaction':np.random.randint(1, 30, n),
    'Churn':np.random.choice(['Yes', 'No'], n),
})
df_clean = pd.DataFrame(data)
```

```python
df_clean
```

```python
import numpy as np
df_dirty = df_clean.copy()

# Missing values (25%)
for col in ["Age", "Usage Frequency", "Support Calls", "Payment Delay", "Last Interaction"]:
    mask = np.random.random(len(df_dirty)) < 0.25
    df_dirty.loc[mask, col] = None

# Total Spend outliers (5%)
mask = np.random.random(len(df_dirty)) < 0.05
df_dirty.loc[mask, "Total Spend"] = \
    np.random.uniform(100, 1000, mask.sum())
```

```python
df = pd.read_csv(r"Desktop/customerchurn_dirty.csv")
df
```

```python
df.dtypes
```

```python
df = pd.concat([df, df.iloc[0:1]])  # demo duplicate
print("Duplicates before drop:", df.duplicated().sum(), "\n")
df.drop_duplicates(inplace=True)
print("After dropping duplicates:\n", df, "\n")
```

```python
print("Missing Values:\n", df.isnull().sum(), "\n")
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Tenure'] = df['Tenure'].replace({'5-':5, '4-':4, '9-':9, '7-':7})
df['Gender'] = df['Gender'].replace({'F': 'Female', 'M': 'Male', 'Fem': 'Female', 'm': 'Male', 'f':'Female'})
df['Usage Frequency'] = df['Usage Frequency'].fillna(df['Usage Frequency'].mean())
df['Support Calls'] = df['Support Calls'].fillna(df['Support Calls'].mean())
df['Payment Delay'] = df['Payment Delay'].fillna(df['Payment Delay'].mean())
df['Subscription Type'] = df['Subscription Type'].replace({'BASIC': 'Basic', 'standard': 'Standard', 'Prem': 'Premium'})
df['Total Spend'] = df['Total Spend'].fillna(df['Total Spend'].median())
df['Last Interaction'] = df['Last Interaction'].fillna(df['Last Interaction'].median())
df['Churn'] = df['Churn'].replace({'Y': 'Yes', 'N': 'No'})
print("After fillna & replace:\n", df, "\n")
```

```python
Q1 = df['Total Spend'].quantile(0.25)
Q3 = df['Total Spend'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df['Total Spend'] = np.where(df['Total Spend'] > upper, upper,
                                 np.where(df['Total Spend'] < lower, lower, df['Total Spend']))
print("After IQR Outlier Capping:\n", df[['Total Spend']], "\n")
```
